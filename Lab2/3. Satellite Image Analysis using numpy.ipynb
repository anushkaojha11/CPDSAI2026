#### **2. Basic Image Analysis**
Now let's find out about our image's data. 
- Check the image's dimensions using the `.shape` attribute.
- Check the data type of the pixels using the `.dtype` attribute.

```python
image_shape = image.shape
image_dtype = image.dtype

print(f"Image Shape (Height, Width, Channels): {image_shape}")
print(f"Image Data Type: {image_dtype}")
```

#### **3. Image Transformations**
Transform the image by manipulating its pixel data. 

**Task 3.1: Invert the Image (Negative)**
Create a new array that is an inversion of the original image.

Inverting an image means flipping every pixel to its opposite brightness/color, what was bright becomes dark, what was dark becomes bright, and colors flip to their "opposite" on the color wheel.

- Subtract every pixel value from 255.

```python
# Subtract every pixel value from 255 to invert the colors
inverted_image = 255 - image

plt.imshow(inverted_image)
plt.title('Inverted Image')
plt.show()
```

**Task 3.2: Convert to Grayscale**
Convert the color image to a single-channel grayscale image.  
A simple way is to average the red, green, and blue channels for each pixel. (image.mean)

## Note 

```python
sample = np.array([[1, 2, 3],
                    [4, 5, 6]])   # shape (2, 3)

print(sample.mean())          # 3.5   → average of ALL elements, collapses everything to one number
print(sample.mean(axis=0))    # [2.5 3.5 4.5]  → average DOWN each column (collapses rows)
print(sample.mean(axis=1))    # [2. 5.]        → average ACROSS each row (collapses columns)
```

For grayscale conversion, we want: for every pixel, average its R, G, and B together into one number. That means we want to collapse only the channel axis, axis 2, while keeping height and width fully intact.

```python
# Calculate mean at the axis 2 (height, width, channels)
grayscale_image = image.mean(axis=2)

print(f"Shape of Original image: {image.shape}")
print(f"Shape of Grayscale image: {grayscale_image.shape}")
```

```python
# Display the grayscale image
plt.imshow(grayscale_image, cmap='gray') # cmap='gray' for dispalying black and white image.
plt.title('Grayscale Image')
plt.show()
```

#### **4. More Transformation: Changing colors in a Range**
We can also use a range to change the pixel values.  
As an example, let's set the green layer for rows 200 to 800 to full intensity.

```python
# Display the original image
plt.imshow(image)
plt.title('Original Image')
plt.show()
```

**Task 4.1: For rows 200 to 800, set Green layer to full intensity 255**
- Hint: Your image has three color channels. You can access green channel using slicing like `image[:, :, 1]`.

```python
# Make a copy of the original image to manipulate
manipulated_image = image.copy()

# For rows 200 to 800, set Green layer to full intensity (255)
manipulated_image[200:800, :, 1] = 255

plt.imshow(manipulated_image)
plt.title('Manipulated Image (Green Boosted, Rows 200-800)')
plt.show()
```

**Task 4.4: Masking**  
Let's mask pixels with value lower than 200 and make them 0 to remove these faint colors.

```python
# Print a few pixels from the original image, for reference
print(image[0:3, 0:3, :])
```

```python
# Make a copy of the original image to manipulate
manipulated_image = image.copy()

# Mask pixels with value lower than 120, and set them to 0 (black)
mask = manipulated_image < 200
manipulated_image[mask] = 0

# Print the same pixels again, now manipulated
print(manipulated_image[0:3, 0:3, :])
```

```python
plt.imshow(image)
plt.title('Original Image')
plt.show()

plt.imshow(manipulated_image)
plt.title('Manipulated Image (Masked < 200)')
plt.show()
```

**Task 4.5: Processing RED pixels**   
Red pixels tell us about the height.  
Let us try to highlight all the high altitude areas.  
We will detect low intensity red pixels and mute them down as well.

```python
# Make a copy of the original image to manipulate
manipulated_image = image.copy()

# Mask Red pixels with value less than 150, and tone them down to 0
# Recall: channel index 0 = Red, 1 = Green, 2 = Blue
red_mask = manipulated_image[:, :, 0] < 150
manipulated_image[red_mask] = 0

plt.imshow(image)
plt.title('Original Image')
plt.show()

plt.imshow(manipulated_image)
plt.title('Manipulated Image (Red < 150 muted)')
plt.show()
```
