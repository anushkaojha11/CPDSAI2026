## RGB channels 

An RGB channel has three channels: red, green and blue. 


![Screenshot](rgb.png)

| Pixel  | R   | G   | B   |
|--------|-----|-----|-----|
| (0,0)  | 102 | 179 | 92  |
| (0,1)  | 14  | 106 | 71  |
| (0,2)  | 188 | 20  | 102 |
| (1,0)  | 121 | 210 | 214 |
| (1,1)  | 74  | 202 | 87  |
| (1,2)  | 116 | 99  | 103 |
| (2,0)  | 151 | 130 | 149 |
| (2,1)  | 52  | 1   | 87  |
| (2,2)  | 235 | 157 | 37  |

```python
#display function to display the images

def show_image_channels(image, title_prefix=""):
    """
    Displays the full RGB image plus its Red, Green, and Blue channels
    separately, with pixel values overlaid on each.
    """
    channels = [
        (None, " Image", None),
        (0, "Red Channel", "Reds"),
        (1, "Green Channel", "Greens"),
        (2, "Blue Channel", "Blues"),
    ]

    plt.figure(figsize=(10, 8))

    for idx, (channel_index, name, cmap) in enumerate(channels, start=1):
        plt.subplot(2, 2, idx)

        if channel_index is None:
            plt.imshow(image)
            for i in range(image.shape[0]):
                for j in range(image.shape[1]):
                    r, g, b = image[i, j, 0], image[i, j, 1], image[i, j, 2]
                    plt.text(j, i, f'R:{r}\nG:{g}\nB:{b}', ha='center', va='center',
                             color='white', fontsize=8,
                             bbox=dict(facecolor='black', alpha=0.5, edgecolor='none', boxstyle='round,pad=0.2'))
        else:
            plt.imshow(image[:, :, channel_index], cmap=cmap)
            for i in range(image.shape[0]):
                for j in range(image.shape[1]):
                    val = image[i, j, channel_index]
                    plt.text(j, i, str(val), ha='center', va='center',
                             color='white', fontsize=12,
                             bbox=dict(facecolor='black', alpha=0.5, edgecolor='none', boxstyle='round,pad=0.2'))

        plt.title(f"{title_prefix}{name}")
        plt.axis("off")

    plt.tight_layout()
    plt.show()
```

Observe the original image

```python
show_image_channels(image, "Original ")
```

## Masking and manipulating the image

```python
# Create a boolean mask: True where the Red channel > 150
mask = image[:, :, 0] > 150
print("Mask (True = Red > 150):")
print(mask)
```

```python
# Use the mask to see which pixels match
print("\nSelected pixels (Red > 150):")
print(image[mask])
```

```python
# Manipulate those pixels — set them to black
image[mask] = [0, 0, 0]

print("\nUpdated image:")
print(image)
```

```python
show_image_channels(image, "Changed ")
```
