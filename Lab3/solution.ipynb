### Troubleshooting: `WebDriverException: Can not connect to the Service`

If you're on a **Mac with Apple Silicon (M1/M2/M3/M4)**, you may see this error when running the cell above. This happens because macOS's kernel refuses to run the downloaded `chromedriver` binary unless it has a valid code signature — and freshly downloaded binaries sometimes don't have one that macOS accepts.

**How to tell if this is your issue:** open Terminal and run the chromedriver binary directly using the exact path from your error message:
```bash
"/path/from/your/error/message/chromedriver" --version
```
If Terminal responds with `zsh: killed`, this is the same issue — the fix below will resolve it.

**Fix:** apply a local ad-hoc code signature to the binary:
```bash
codesign --force --deep --sign - "/path/from/your/error/message/chromedriver"
```
Then re-run the version check above to confirm it now prints a real version string, and re-run the cell above.

**Note:** this fix is tied to *this specific downloaded file*. If Chrome updates itself later and `ChromeDriverManager()` downloads a newer chromedriver version, you may need to repeat this fix on the new file.
